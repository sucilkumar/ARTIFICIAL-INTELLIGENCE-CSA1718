```python
import heapq

class PuzzleState:
    def __init__(self, board, parent=None, move="", g=0):
        self.board = board  # 1D tuple of length 9
        self.parent = parent
        self.move = move
        self.g = g  # Path cost from start to current node
        self.h = self.calculate_manhattan_distance()  # Heuristic cost
        self.f = self.g + self.h  # Total estimated cost

    def __lt__(self, other):
        return self.f < other.f

    def calculate_manhattan_distance(self):
        distance = 0
        # Goal positions for each number 1-8. 0 is the empty tile (ignored in distance)
        goal_positions = {1: (0, 1), 2: (0, 2), 3: (1, 0), 4: (1, 1), 5: (1, 2), 6: (2, 0), 7: (2, 1), 8: (2, 2)}

        for i, val in enumerate(self.board):
            if val != 0:
                current_row, current_col = divmod(i, 3)
                goal_row, goal_col = goal_positions[val]
                distance += abs(current_row - goal_row) + abs(current_col - goal_col)
        return distance

    def get_neighbors(self):
        neighbors = []
        zero_index = self.board.index(0)
        row, col = divmod(zero_index, 3)

        # Possible directions of movements for the blank (0) space
        moves = {
            "Up": (-1, 0),
            "Down": (1, 0),
            "Left": (0, -1),
            "Right": (0, 1)
        }

        for dir_name, (dr, dc) in moves.items():
            new_row, new_col = row + dr, col + dc
            if 0 <= new_row < 3 and 0 <= new_col < 3:
                new_zero_index = new_row * 3 + new_col
                # Swap the empty tile with the adjacent tile
                new_board = list(self.board)
                new_board[zero_index], new_board[new_zero_index] = new_board[new_zero_index], new_board[zero_index]
                neighbors.append(PuzzleState(tuple(new_board), self, dir_name, self.g + 1))

        return neighbors

def solve_puzzle(start_board):
    goal_board = (1, 2, 3, 4, 5, 6, 7, 8, 0)

    # Check solvability based on inversions
    def is_solvable(board):
        inversions = 0
        flat_board = [x for x in board if x != 0]
        for i in range(len(flat_board)):
            for j in range(i + 1, len(flat_board)):
                if flat_board[i] > flat_board[j]:
                    inversions += 1
        return inversions % 2 == 0

    if not is_solvable(start_board):
        return None

    start_state = PuzzleState(start_board)
    open_set = []
    heapq.heappush(open_set, start_state)
    closed_set = set()

    while open_set:
        current_state = heapq.heappop(open_set)

        if current_state.board == goal_board:
            # Reconstruct path
            path = []
            while current_state.parent:
                path.append((current_state.move, current_state.board))
                current_state = current_state.parent
            path.reverse()
            return path

        closed_set.add(current_state.board)

        for neighbor in current_state.get_neighbors():
            if neighbor.board in closed_set:
                continue
            heapq.heappush(open_set, neighbor)

    return None

def print_board(board):
    for i in range(0, 9, 3):
        print(f"{board[i]} {board[i+1]} {board[i+2]}")
    print()
```

```python
# Define the initial state (0 represents the empty block)
# 1 2 3
# 0 4 6
# 7 5 8
initial_state = (1, 2, 3, 0, 4, 6, 7, 5, 8)

print("Initial State:")
print_board(initial_state)

solution = solve_puzzle(initial_state)

if solution is not None:
    print(f"Solvable in {len(solution)} moves!\n")
    current_board = initial_state
    for step, (move, board) in enumerate(solution, 1):
        print(f"Step {step}: Move blank space '{move}'")
        print_board(board)
else:
    print("This puzzle configuration is not solvable.")
```
